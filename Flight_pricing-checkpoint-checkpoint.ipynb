#### Data preprocessing

```python
df["Journey_day"]=pd.to_datetime(df.Date_of_Journey,format="%d/%m/%Y").dt.day
```

```python
df["Journey_month"]=pd.to_datetime(df.Date_of_Journey,format="%d/%m/%Y").dt.month
```

```python
df.head()
```

```python
# deoping the data of journey , because it is no more need 
df.drop(["Date_of_Journey"],axis=1,inplace=True) # axis =1 {drop the  column and its observation(record) }
```

```python
df.head()
```

```python
# extracting only the hour from "Dep_Time"
df["Dep_Hour"]=pd.to_datetime(df["Dep_Time"],format='%H:%M').dt.hour


# extracting only the minute from "Dep_Time"
df["Dep_MIN"]=pd.to_datetime(df["Dep_Time"],format="%H:%M").dt.minute
```

```python
# droping the dep_time, specfice info is extracted
df.drop(["Dep_Time"],axis=1,inplace=True)
```

```python
df.head()
```

```python
# Automatically handle mixed datetime formats in the column
arrival_dt = pd.to_datetime(df["Arrival_Time"], format="mixed") # the formate in the "Arrival_time was mixed in few column "

# Extract only the hour and minute
df["Arrival_Hour"] = arrival_dt.dt.hour
df["Arrival_Min"] = arrival_dt.dt.minute
```

```python
#droping the arrival time
df.drop(["Arrival_Time"],axis=1,inplace=True)
```

```python
df.head()
```

```python

# Assigning and converting Duration column into list
duration = list(df["Duration"])

for i in range(len(duration)):
    if len(duration[i].split()) != 2:
        if "h" in duration[i]:
            duration[i] = duration[i].strip() + " 0m"
        else:
            duration[i] = "0h "+ duration[i]

duration_hour = []
duration_min = []
for i in range(len(duration)):
    duration_hour.append(int(duration[i].split(sep = "h")[0])) 
    duration_min.append(int(duration[i].split(sep = "m")[0].split()[-1]))

```

```python
df["Duration_hour"]=duration_hour
df["Duration_min"]=duration_min
```

```python
df.drop(["Duration"],axis=1,inplace=True)
```

```python
df.head()
```

##### Model Training

```python
from sklearn.model_selection import train_test_split,cross_val_score
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=32)
```

```python
x_train.shape,y_train.shape,x_test.shape,y_test.shape
```

```python
y_train_log = np.log1p(y_train)
```

```python
y_test_log=np.log1p(y_test)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
gbr_m=GradientBoostingRegressor()
```

```python
# model fit 
gbr_m.fit(x_train,y_train_log)
```

```python
y_pred_log=gbr_m.predict(x_test)
```

##### Model  Evaluation

```python
print(gbr_m.score(x_train,y_train_log))
print(gbr_m.score(x_test,y_test_log))
```

```python
r2=r2_score(y_test_log,y_pred_log)
print("R2 score: ",r2)
```

```python
y_predi=np.expm1(y_pred_log)
```

```python
y_testi=np.expm1(y_test_log)
```

```python
r2=r2_score(y_testi,y_predi)
print("R2 score: ",r2)
```

```python
mae= mean_absolute_error(y_test,y_predi)
print("MAE: ",mae)
```

```python
mse = mean_squared_error(y_test, y_predi)
print("MSE:", mse)
```

#### challenges 
Working on this flight price prediction project came with a few challenge. Handling all the categorical text data like airlines and routes without bloating the dataset was definitely tricky at first. The ticket prices also had a huge range—from cheap economy seats to expensive last-minute fares—which created a heavy "Righr skew" that threw off the model until I used a log transformation. Parsing the raw timestamps and durations into clean numbers took some extra Pandas work, and figuring out how to tune like(max_depth or learning rate) the model.
